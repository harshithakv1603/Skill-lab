```python
import os
from getpass import getpass

os.environ["GEMINI_API_KEY"] = getpass("Enter Gemini API Key: ")

print("Gemini API key configured successfully!")
```

```python
import asyncio
from google import genai
from google.genai import types
from playwright.async_api import async_playwright

# Initialize Gemini client
client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

print("All libraries imported successfully!")
```

```python
async def start_browser():

    global pw, browser, context, page

    pw = await async_playwright().start()

    browser = await pw.chromium.launch(
        headless=True
    )

    context = await browser.new_context()

    page = await context.new_page()

    print("Chrome browser is ready!")

await start_browser()
```

```python
print("Browser connected:", browser.is_connected())
```

```python
async def open_amazon():

    print("Opening Amazon India...")

    await page.goto(
        "https://www.amazon.in",
        wait_until="domcontentloaded",
        timeout=60000
    )

    print("Amazon opened successfully!")

    print("Website title:", await page.title())
    print("Website URL:", page.url)


await open_amazon()
```

```python
print("Browser connected:", browser.is_connected())
print("Current URL:", page.url)
```

```python
async def search_product(keyword: str):

    print(f"Searching for: {keyword}")

    try:

        search_box = page.locator(
            "#twotabsearchtextbox"
        )

        await search_box.wait_for(
            state="visible",
            timeout=15000
        )

        await search_box.fill(keyword)

        await page.locator(
            "#nav-search-submit-button"
        ).click()

        await page.wait_for_selector(
            "[data-component-type='s-search-result']",
            timeout=30000
        )

        print("Product search completed!")

        return True

    except Exception as e:

        print("Search failed:", str(e))

        return False


result = await search_product(
    "wireless mouse under 1000"
)

print("Search status:", result)
```

```python
print(
    "Search box count:",
    await page.locator(
        "#twotabsearchtextbox"
    ).count()
)

print("Current URL:", page.url)
```

```python
print("Current URL:", page.url)
print("Page title:", await page.title())
print("Browser connected:", browser.is_connected())
```

```python
try:
    await browser.close()
except:
    pass

try:
    await pw.stop()
except:
    pass

print("Old browser session cleaned!")
```

```python
import asyncio
from playwright.async_api import async_playwright

# Start a fresh Playwright session
pw = await async_playwright().start()

# Launch browser
browser = await pw.chromium.launch(
    headless=True
)

# Create new context
context = await browser.new_context()

# Create new page
page = await context.new_page()

print("Playwright started successfully!")
print("Browser connected:", browser.is_connected())
print("Page closed:", page.is_closed())
```

```python
await page.goto(
    "https://www.amazon.in",
    wait_until="domcontentloaded",
    timeout=60000
)

print("Amazon URL:", page.url)
print("Amazon title:", await page.title())
```

```python
await page.wait_for_timeout(3000)

captcha = await page.locator(
    "form[action*='validateCaptcha']"
).count()

search = await page.locator(
    "#twotabsearchtextbox"
).count()

print("CAPTCHA count:", captcha)
print("Search box count:", search)
```

```python
print((await page.locator("body").inner_text())[:2000])
```

```python
print("Page title:", await page.title())
print("Current URL:", page.url)
```

```python
print(
    "Continue shopping buttons:",
    await page.get_by_text(
        "Continue shopping",
        exact=True
    ).count()
)
```

```python
button = page.get_by_text(
    "Continue shopping",
    exact=True
)

print("Visible:", await button.is_visible())
print("Enabled:", await button.is_enabled())
print("Tag name:", await button.evaluate("(el) => el.tagName"))
```

```python
await button.click(timeout=10000)

await page.wait_for_timeout(5000)

print("Current URL:", page.url)
print("Page title:", await page.title())
```
